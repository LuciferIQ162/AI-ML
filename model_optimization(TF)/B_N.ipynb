```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, Flatten, BatchNormalization
```

```python
#define a simple model with batch Normalization
model = Sequential([
    Flatten(input_shape=(28, 28)),
    Dense(128, activation='relu'),
    BatchNormalization(),#Batch Normalization layer added
    Dense(10, activation='relu'),
])
```

# (optional) mixed_precision_training

```python
from tensorflow.keras import mixed_precision
#enable mixed precision training
policy = mixed_precision.Policy('mixed_float16')
mixed_precision.set_global_policy(policy)
```

# model pruning

```python
# model pruning
import tensorflow as tf
import tensorflow_model_optimization as tfmot

# Define pruning schedule (adjust end_step after knowing steps_per_epoch * epochs)
pruning_params = {
    'pruning_schedule': tfmot.sparsity.keras.PolynomialDecay(
        initial_sparsity=0.0,
        final_sparsity=0.5,
        begin_step=0,
        end_step=2000  # update later if needed
    )
}

model_pruned = tfmot.sparsity.kerasprune_low_magnitude(model, **pruning_params)
```

Quantization
# knowledge distillation

```python
#example code for quantization
#example code for post-training Quantization
converter = tf.lite.TFLiteConverter.from_keras_model(model)
converter.optimizations = [tf.lite.Optimize.DEFAULT]
quantized_model = converter.convert()

#save the quantized Model
with open('quantized_model.tflite', 'wb') as f:
    f.write(quantized_model)
```
