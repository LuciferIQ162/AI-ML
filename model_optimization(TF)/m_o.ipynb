Weight initialization: Example

```python
from tensorflow.keras.initializers import HeNormal
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.math import exp
```

```python
#define the model with "he" initialization
model = Sequential([
    Flatten(input_shape=(28, 28)),
    Dense(128, activation='relu', kernel_initializer=HeNormal()),#he initialization applied
    Dense(10, activation='softmax')
])
```

# learning rate scheduling

```python
#load and preprocess data
from tensorflow.keras.datasets import mnist
from tensorflow.keras.utils import to_categorical
(x_train, y_train), (x_val, y_val) = mnist.load_data()
x_train = x_train.astype('float32')/ 255.0
x_val = x_val.astype('float32')/ 255.0

#reshaping the input data(if needed)
x_train = x_train.reshape(-1, 28, 28)
x_val = x_val.reshape(-1, 28, 28)
```

```python
from tensorflow.keras.callbacks import LearningRateScheduler
from tensorflow.keras.optimizers import Adam

# Exponential decay after epoch 10 (using e^-0.1)
def scheduler(epoch, lr):
    return float(lr) if epoch < 10 else float(lr * 2.718281828**(-0.1))

lr_scheduler = LearningRateScheduler(scheduler, verbose=1)
```

```python
#model training

#compile the model
model.compile(optimizer=Adam(),
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

history = model.fit(
    x_train, y_train,
    validation_data=(x_val, y_val),
    epochs=20,
    callbacks=[lr_scheduler]
)
```
