```python
import tensorflow as tf
from tensorflow.keras.datasets import mnist, fashion_mnist
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
#load and preprocess the data mnist
(x_train, _), (_, _) = mnist.load_data()
x_train = x_train.astype('float32') / 255.0
x_train = x_train.reshape(-1, 28 * 28)
```

```python
#load and preprocess the data mnist
(x_train, _), (_, _) = fashion_mnist.load_data()
x_train = x_train.astype('float32') / 255.0
x_train = x_train.reshape(-1, 28 * 28)
```

```python
#apply kmeans clustering
kmeans = KMeans(n_clusters=10)
kmeans.fit(x_train)
labels = kmeans.labels_
```

```python
#displey some clustered images
def display_clustered_images(x_train, labels):
    plt.figure(figsize=(10,10))
    for i in range(10):
        idxs = np.where(labels == i)[0]
        for j in range(10):
                plt_idx = i * 10 + j + 1
                plt.subplot(10, 10, plt_idx)
                plt.imshow(x_train[idxs[j]].reshape(28, 28), cmap='bone')
                plt.axis('off')
    plt.show()

display_clustered_images(x_train, labels)
```
